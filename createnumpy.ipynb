```python
import numpy as np

arr = np.array([1,3,4,4,5])
print(arr)
```

```python
intZeroArr = np.zeros((5,4), dtype=int)
intZeroArr2 = np.zeros((5,4), dtype = int)

# print(intZeroArr)
# print(intZeroArr2)
print(f"Is Both Array Equal:: {np.array_equal(intZeroArr, intZeroArr2)}")
```

```python
# Use of np.all
print(np.all(intZeroArr == intZeroArr2))
```

```python
floatArr = np.array([0.1 + 0.2])
print(floatArr)

floatArr2 = np.array([0.3])
print(floatArr2)

print(f"is Both Array Equal? {np.array_equal(floatArr,floatArr2)}")

print(f"is np closeAll ? {np.allclose(floatArr,floatArr2)}")

print(np.all(floatArr == floatArr2))
```
